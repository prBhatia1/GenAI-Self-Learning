```python
from sentence_transformers import SentenceTransformer
```

```python
open_source_embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
text = "Machine"
```

```python
embedding = open_source_embedding_model.encode(text)
```
